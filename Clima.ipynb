```python
def cal_temperatura (T):
    if T > 26:
        return "quente"
    elif T <= 26 and T >= 20:
        return "ameno"
    elif T <= 20 and T > 15:
        return "Fresco"
    elif T <= 15:
        return "frio"

def cal_umidade (U):
    if U >= 95:
        return "Elevada"
    elif U < 95 and U >= 70:
        return "Normal"
    elif U < 70:
        return "Fraca"


def cal_vento (V):
    if V == 0:
        return "fraco"
    elif V == 1: 
        return "forte"

    
```
